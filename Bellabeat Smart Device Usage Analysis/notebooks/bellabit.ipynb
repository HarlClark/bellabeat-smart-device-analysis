```python
daily_activity_clean = d_a.copy()
sleep_day_clean = s_d.copy()
hourly_steps_clean = h_s.copy()
hourly_intensities_clean = h_i.copy()
```

```python
sleep_day_clean = sleep_day_clean.drop_duplicates()
```

```python
print("Sleep duplicates after cleaning:",
      sleep_day_clean.duplicated().sum())
```

```python
daily_activity_clean["ActivityDate"] = pd.to_datetime(
    daily_activity_clean["ActivityDate"]
)
```

```python
sleep_day_clean["SleepDay"] = pd.to_datetime(
    sleep_day_clean["SleepDay"]
)
```

```python
hourly_steps_clean["ActivityHour"] = pd.to_datetime(
    hourly_steps_clean["ActivityHour"]
)
```

```python
hourly_intensities_clean["ActivityHour"] = pd.to_datetime(
    hourly_intensities_clean["ActivityHour"]
)
```

```python
daily_activity_clean["DayOfWeek"] = (
    daily_activity_clean["ActivityDate"]
    .dt.day_name()
)
```

```python
sleep_day_clean["DayOfWeek"] = (
    sleep_day_clean["SleepDay"]
    .dt.day_name()
)
```

```python
sleep_day_clean["TotalSleepHours"] = (
    sleep_day_clean["TotalMinutesAsleep"] / 60
)
```

```python
sleep_day_clean["TimeInBedHours"] = (
    sleep_day_clean["TotalTimeInBed"] / 60
)
```

```python
daily_activity_clean["TotalActiveMinutes"] = (
    daily_activity_clean["VeryActiveMinutes"]
    + daily_activity_clean["FairlyActiveMinutes"]
    + daily_activity_clean["LightlyActiveMinutes"]
)
```

```python
daily_activity_clean["SedentaryPercentage"] = (
    daily_activity_clean["SedentaryMinutes"] / (24 * 60)
) * 100
```

```python
def classify_activity(steps):
    if steps < 5000:
        return "Low"
    elif steps < 10000:
        return "Moderate"
    else:
        return "High"
```

```python
daily_activity_clean["ActivityLevel"] = (
    daily_activity_clean["TotalSteps"]
    .apply(classify_activity)
)
```

```python
daily_activity_clean[
    [
        "TotalSteps",
        "Calories",
        "VeryActiveMinutes",
        "FairlyActiveMinutes",
        "LightlyActiveMinutes",
        "SedentaryMinutes"
    ]
].describe()
```

```python
activity_summary = daily_activity_clean[
    [
        "TotalSteps",
        "Calories",
        "VeryActiveMinutes",
        "FairlyActiveMinutes",
        "LightlyActiveMinutes",
        "SedentaryMinutes"
    ]
].mean()

print(activity_summary)
```

```python
weekday_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

weekday_activity = (
    daily_activity_clean
    .groupby("DayOfWeek")["TotalSteps"]
    .mean()
    .reindex(weekday_order)
)

print(weekday_activity)
```

```python
plt.figure(figsize=(10, 5))

weekday_activity.plot(kind="bar")

plt.title("Average Daily Steps by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Average Steps")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sleep_summary = sleep_day_clean[
    ["TotalSleepHours", "TimeInBedHours"]
].mean()

print(sleep_summary)
```

```python
weekday_sleep = (
    sleep_day_clean
    .groupby("DayOfWeek")["TotalSleepHours"]
    .mean()
    .reindex(weekday_order)
)

print(weekday_sleep)
```

```python
plt.figure(figsize=(10, 5))

weekday_sleep.plot(kind="bar")

plt.title("Average Sleep Duration by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Average Sleep (Hours)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
hourly_steps_clean["Hour"] = (
    hourly_steps_clean["ActivityHour"].dt.hour
)
```

```python
hourly_activity = (
    hourly_steps_clean
    .groupby("Hour")["StepTotal"]
    .mean()
)

print(hourly_activity)
```

```python
plt.figure(figsize=(12, 5))

hourly_activity.plot(kind="line", marker="o")

plt.title("Average Steps by Hour of Day")
plt.xlabel("Hour")
plt.ylabel("Average Steps")

plt.xticks(range(24))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
hourly_intensities_clean["Hour"] = (
    hourly_intensities_clean["ActivityHour"].dt.hour
)
```

```python
hourly_intensity = (
    hourly_intensities_clean
    .groupby("Hour")["TotalIntensity"]
    .mean()
)

print(hourly_intensity)
```

```python
plt.figure(figsize=(12, 5))

hourly_intensity.plot(kind="line", marker="o")

plt.title("Average Activity Intensity by Hour")
plt.xlabel("Hour")
plt.ylabel("Average Intensity")

plt.xticks(range(24))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
activity_for_sleep = daily_activity_clean[
    ["Id", "ActivityDate", "TotalSteps",
     "VeryActiveMinutes", "FairlyActiveMinutes",
     "LightlyActiveMinutes", "SedentaryMinutes"]
].copy()

activity_for_sleep.rename(
    columns={"ActivityDate": "Date"},
    inplace=True
)

sleep_for_merge = sleep_day_clean[
    ["Id", "SleepDay", "TotalSleepHours"]
].copy()

sleep_for_merge.rename(
    columns={"SleepDay": "Date"},
    inplace=True
)
```

```python
activity_sleep = pd.merge(
    activity_for_sleep,
    sleep_for_merge,
    on=["Id", "Date"],
    how="inner"
)
```

```python
print(activity_sleep.shape)
activity_sleep.head()
```

```python
correlation = activity_sleep[
    [
        "TotalSteps",
        "VeryActiveMinutes",
        "FairlyActiveMinutes",
        "LightlyActiveMinutes",
        "SedentaryMinutes",
        "TotalSleepHours"
    ]
].corr()

print(correlation)
```

```python
print(
    "Steps vs Sleep:",
    activity_sleep["TotalSteps"]
    .corr(activity_sleep["TotalSleepHours"])
)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    activity_sleep["TotalSteps"],
    activity_sleep["TotalSleepHours"]
)

plt.title("Daily Steps vs Sleep Duration")
plt.xlabel("Total Steps")
plt.ylabel("Sleep Duration (Hours)")

plt.tight_layout()
plt.show()
```
