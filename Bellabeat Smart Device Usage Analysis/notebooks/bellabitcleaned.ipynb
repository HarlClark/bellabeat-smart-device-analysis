```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

daily_activity = pd.read_csv("Cleaned/dailyActivity_cleaned.csv")
sleep_day = pd.read_csv("Cleaned/sleepDay_cleaned.csv")
hourly_steps = pd.read_csv("Cleaned/hourlySteps_cleaned.csv")
hourly_intensities = pd.read_csv("Cleaned/hourlyIntensities_cleaned.csv")
```

```python
daily_activity[
    [
        "TotalSteps",
        "TotalDistance",
        "Calories",
        "VeryActiveMinutes",
        "FairlyActiveMinutes",
        "LightlyActiveMinutes",
        "SedentaryMinutes"
    ]
].describe().round(2)
```

```python
sleep_day[
    [
        "TotalSleepHours",
        "TimeInBedHours"
    ]
].describe().round(2)
```

```python
daily_activity["ActivityLevel"].value_counts()
```

```python
activity_distribution = (
    daily_activity["ActivityLevel"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(activity_distribution)
```

```python
weekday_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

weekday_analysis = (
    daily_activity
    .groupby("DayOfWeek")
    .agg(
        AvgSteps=("TotalSteps", "mean"),
        AvgCalories=("Calories", "mean"),
        AvgSedentaryMinutes=("SedentaryMinutes", "mean"),
        AvgActiveMinutes=("TotalActiveMinutes", "mean")
    )
    .reindex(weekday_order)
    .round(2)
)

weekday_analysis
```

```python
print(
    "Most active day:",
    weekday_analysis["AvgSteps"].idxmax()
)

print(
    "Least active day:",
    weekday_analysis["AvgSteps"].idxmin()
)
```

```python
hourly_analysis = (
    hourly_steps
    .groupby("Hour")
    .agg(
        AvgSteps=("StepTotal", "mean")
    )
)

hourly_analysis
```

```python
peak_hour = hourly_analysis["AvgSteps"].idxmax()

print("Peak activity hour:", peak_hour)
```

```python
print(
    hourly_analysis.sort_values(
        "AvgSteps",
        ascending=False
    ).head(5)
)
```

```python
sleep_analysis = (
    sleep_day
    .agg(
        AvgSleepHours=("TotalSleepHours", "mean"),
        MedianSleepHours=("TotalSleepHours", "median"),
        AvgTimeInBed=("TimeInBedHours", "mean")
    )
)

sleep_analysis.round(2)
```

```python
weekday_sleep = (
    sleep_day
    .groupby("DayOfWeek")
    .agg(
        AvgSleepHours=("TotalSleepHours", "mean"),
        AvgTimeInBed=("TimeInBedHours", "mean")
    )
    .reindex(weekday_order)
    .round(2)
)

weekday_sleep
```

```python
activity_sleep[
    [
        "TotalSteps",
        "VeryActiveMinutes",
        "FairlyActiveMinutes",
        "LightlyActiveMinutes",
        "SedentaryMinutes",
        "TotalSleepHours"
    ]
].corr().round(2)
```
